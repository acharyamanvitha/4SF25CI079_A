Loc

loc is used to select rows and coloumns by label.
you can pass it
-a single label
-a list of labels
-a slice of labels
-a boolean series
-a colon(which indicates all labels)

before the : it is rows and you put index values
-after : it is colomns and you put name


```python
df.loc[0,'Product Name']
```

```python
df.loc[0,:]
```

```python
df.loc[[0,1,2],:]
```

```python
df.loc[0:2,:]
```

```python
df.loc[0:2]
```

Handling missing values

```python
df.isnull().any()
```

```python
df.isnull().sum() #this shows the number of null values in each column of the DataFrame.
```

```python
df_filled=df.fillna(0) #this replaces all the null values in the DataFrame with 0.
df_filled.head(50)
```

```python
df['SALES NEW']=df['Sales']
df.head(5)
```

```python
df['Sales'].mean() #this calculates the mean of the 'Sales' column in the DataFrame.
```

```python
df.fillna(df['SALES NEW'].mean(), inplace=True) #this replaces all the null values in the 'Sales' column with the mean of that column.
```

```python
df['Value_new']=df['Value'].astype(int) #this converts the 'Value' column to integer data type and creates a new column 'Value_new' with the converted values.
df
```

ok so we will get here error if we have any null values in that coloumn because null value donot have any data type

```python
df['New_Value']=df['Value'].fillna(df['Value'].mean()).astype(float)
df.head(5)
```

```python
df['New_Value']=df['Value'].apply(lambda x: x*2) #this multiplies each value in the 'Value' column by 2 and creates a new column 'New_Value' with the results.
df.head(5)
```
