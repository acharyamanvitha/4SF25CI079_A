1. Counting Words in a File
Write a function that reads the contents of a file named `document.txt` and returns the number of words in the file.

```python
def count_words():
    data = pd.read_csv("document.txt", header=None)
    text = " ".join(data[0].astype(str))
    words = text.split()
    return len(words)

print(count_words())
```

2. Finding and Replacing Text
Write a function that finds and replaces all occurrences of a given word in a file named `data.txt` with another word.

```python
def find_replace(old_word, new_word):
    data = pd.read_csv("data.txt", header=None)

    data[0] = data[0].astype(str).str.replace(old_word, new_word)

    data.to_csv("data.txt", index=False, header=False)

find_replace("Python", "Java")
```

3. Reading a File in Reverse
Write a function that reads the contents of a file named `reverse.txt` and prints each line in reverse order.

```python
def reverse_file():
    data = pd.read_csv("reverse.txt", header=None)

    print("Original File:")
    for line in data[0]:
        print(line)

    print("\nReversed File:")
    for line in data[0]:
        print(str(line)[::-1])

reverse_file()
```

4. Counting Lines, Words, and Characters
Write a function that reads the contents of a file named `stats.txt` and returns the number of lines, words, and characters in the file.


```python
def file_stats():
    data = pd.read_csv("document.txt", header=None)

    lines = len(data)
    words = sum(len(str(line).split()) for line in data[0])
    characters = sum(len(str(line)) for line in data[0])

    print("Number of lines:", lines)
    print("Number of words:", words)
    print("Number of characters:", characters)

file_stats()
```

15. Data Aggregation 
Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.

```python
df = pd.DataFrame({
    "Category": np.random.choice(["A", "B", "C"], 10),
    "Value": np.random.randint(1, 101, 10)
})

print("Original DataFrame:")
print(df)

result = df.groupby("Category")["Value"].agg(["sum", "mean"])

print("\nSum and Mean of Value for each Category:")
print(result)
```

17. Merging DataFrames
Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.

```python
df1 = pd.DataFrame({
    "ID": [1, 2, 3, 4],
    "Name": ["Anu", "Rahul", "Meera", "Kiran"]
})

df2 = pd.DataFrame({
    "ID": [1, 2, 3, 4],
    "Marks": [85, 90, 78, 92]
})

print("DataFrame 1:")
print(df1)

print("\nDataFrame 2:")
print(df2)

# Merge using common column
merged_df = pd.merge(df1, df2, on="ID")

print("\nMerged DataFrame:")
print(merged_df)
```

18. MultiIndex DataFrame
Create a Pandas DataFrame with a MultiIndex (hierarchical index). Perform some basic indexing and slicing operations on the MultiIndex DataFrame.

```python
index = pd.MultiIndex.from_tuples([
    ("A", 1),
    ("A", 2),
    ("B", 1),
    ("B", 2)
], names=["Category", "Number"])

df = pd.DataFrame({
    "Value": [10, 20, 30, 40]
}, index=index)

print("MultiIndex DataFrame:")
print(df)

# Basic indexing
print("\nData for Category A:")
print(df.loc["A"])

# Access a specific value
print("\nValue of A, 2:")
print(df.loc[("A", 2)])

# Slicing
print("\nData from A to B:")
print(df.loc["A":"B"])
```

19. Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'. Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.

```python
categories = ["A", "A", "B", "B", "C", "C"]
subcategories = ["X", "Y", "X", "Y", "X", "Y"]

index = pd.MultiIndex.from_arrays(
    [categories, subcategories],
    names=["Category", "SubCategory"]
)

df = pd.DataFrame({
    "Value": np.random.randint(1, 101, 6)
}, index=index)

print("MultiIndex DataFrame:")
print(df)

print("\nSum of Values:")
print(df.groupby(["Category", "SubCategory"])["Value"].sum())
```

20. Applying Functions
Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Apply a function that doubles the values of the DataFrame.

```python
df = pd.DataFrame(
    np.random.randint(1, 11, (5, 3)),
    columns=["A", "B", "C"]
)

print("Original DataFrame:")
print(df)

# Function to double the values
def double(x):
    return x * 2

df = df.map(double)

print("\nDataFrame after doubling:")
print(df)
```

21. Create a Pandas DataFrame with 3 columns and 6 rows filled with random integers. Apply a lambda function to create a new column that is the sum of the existing columns.

```python
df = pd.DataFrame(
    np.random.randint(1, 11, (6, 3)),
    columns=["A", "B", "C"]
)

print("Original DataFrame:")
print(df)

# Create new column using lambda
df["Total"] = df.apply(lambda row: row["A"] + row["B"] + row["C"], axis=1)

print("\nDataFrame after adding Total:")
print(df)
```
