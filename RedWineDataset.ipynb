What does correlation mean?
Correlation tells us how two variables move in relation to each other.

For example, in the wine dataset, we might want to know:
- When alcohol increases, does wine quality tend to increase?
- When volatile acidity increases, does wine quality tend to decrease?

Correlation gives us a numerical measurement of these relationships.

How do we interpret correlation values?
Correlation typically ranges from -1 to +1.

+1
Perfect positive correlation
Both variables move together: when one increases, the other increases proportionally.

0
Zero linear correlation
No linear relationship is detected. A nonlinear relationship may still exist.

-1
Perfect negative correlation
One variable increases as the other decreases, in a perfectly linear relationship.

Pandas calculates a correlation matrix. By default, it uses Pearson correlation, 
which measures linear relationships between numerical columns.

The output looks conceptually like this:

	alcohol	density	quality
alcohol	1.00	-0.50	0.48
density	-0.50	1.00	-0.30
quality	0.48	-0.30	1.00

These numbers are illustrative, not the calculated results from your dataset.
How to read it:
- alcohol vs. quality = 0.48: a positive relationship.
- alcohol vs. density = -0.50: a negative relationship.
- A column correlated with itself always has a correlation of 1.00.

```python
##histogram
for column in df.columns:
    sns.histplot(df[column],kde=True)
```

Why do this during EDA? it helps identify:

distribution: where most values are concentrated

skewness:whether values are concentrated on one side

outliers: values that appear far from the main distribuation

spread: whether values are tightly clustered or widely distributed
